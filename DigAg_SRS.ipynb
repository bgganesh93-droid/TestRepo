## 2. Introduction

### 2.1 Purpose
This document defines the functional and non-functional requirements for a system that automates aerial data collection using Parrot Anafi drones and streamlines the transfer of captured imagery from field devices to high-performance computing (HPC) environments. The goal is to minimize manual intervention, ensure efficient mission planning, and trigger downstream ML workflows automatically and reliably.

### 2.2 Scope
The system comprises two main components:

- **Path Planner Module**: Generates optimized waypoint grids for drone missions based on user-defined field dimensions and drone/camera configurations.
- **Transfer Module**: Monitors the mission image folder, batches and uploads images asynchronously to an HPC system (e.g., OSC via Tapis), and initiates training or inference jobs.

The solution is intended for use in edge environments (e.g., agricultural fields) with limited computing resources and unreliable network connectivity.

| Waypoint | GPS coordinate that a drone follows during a flight mission |
| OpenPass | Open-source drone mission controller used with Parrot Anafi |
| Anafi | A lightweight quadcopter drone by Parrot, used for aerial image capture |

## 3. Overall Description

### 3.1 Product Perspective
The system is part of a larger field sensing and cloud pipeline architecture, facilitating:

- Efficient **pre-flight mission planning**
- **Real-time monitoring** of image generation
- **Reliable image transfer** to remote servers (e.g., OSC)
- **Triggering of downstream AI jobs** (e.g., segmentation or classification)

The architecture assumes no AI processing is performed on the drone; all inference/training occurs on the HPC backend.

## 4. Functional Requirements

### 4.1 Path Planner Module

#### 4.1.1 Plot Input and Configuration

**FR-PP-001: Rectangular Plot Definition**
- The system shall accept user input defining rectangular field plots via four corner coordinates (latitude, longitude)
- Priority: High
- Dependencies: None

**FR-PP-002: Alternative Plot Input Methods**
- The system shall support plot definition through:
  - Manual coordinate entry (decimal degrees)
  - GeoJSON file import
  - Interactive map-based selection (optional GUI)
- Priority: Medium
- Dependencies: FR-PP-001

**FR-PP-003: Plot Validation**
- The system shall validate that input coordinates form a valid rectangle within ±10° of perpendicularity
- The system shall reject plots smaller than 10m × 10m or larger than 5km × 5km
- The system shall display clear error messages for invalid inputs
- Priority: High
- Dependencies: FR-PP-001

**FR-PP-004: Flight Parameter Configuration**
- The system shall allow users to specify:
  - Flight altitude (AGL): 10m to 120m
  - Camera angle: 0° to 90° from nadir
  - Image overlap: forward (60-90%) and side (40-80%)
  - Target Ground Sample Distance (GSD): 0.5cm to 10cm per pixel
  - Flight speed: 1m/s to 10m/s
- Priority: High
- Dependencies: None

#### 4.1.2 Waypoint Generation

**FR-PP-005: Flight Path Optimization**
- The system shall compute optimal waypoint spacing based on:
  - Camera field of view (FOV) for Parrot Anafi
  - Specified flight altitude
  - Target image overlap percentages
  - Target GSD requirements
- Priority: High
- Dependencies: FR-PP-004

**FR-PP-006: Lawnmower Pattern Generation**
- The system shall generate parallel transect paths covering the entire plot area
- The system shall minimize total flight distance while maintaining coverage requirements
- The system shall include 180° turn waypoints at row endpoints
- Priority: High
- Dependencies: FR-PP-005

**FR-PP-007: Waypoint Ordering**
- The system shall order waypoints to minimize flight time
- The system shall include takeoff point as first waypoint and landing point as final waypoint
- The system shall support both North-South and East-West primary flight directions
- Priority: Medium
- Dependencies: FR-PP-006

**FR-PP-008: Altitude Adjustment**
- The system shall maintain consistent altitude (AGL) across all waypoints
- The system shall support optional terrain-following mode if elevation data is provided
- Priority: Medium
- Dependencies: FR-PP-005

#### 4.1.3 Mission Manifest Generation

**FR-PP-009: Mission File Creation**
- The system shall generate mission manifest files compatible with OpenPass format
- The manifest shall include:
  - Complete ordered waypoint list (lat, lon, altitude)
  - Flight parameters (speed, camera settings)
  - Estimated flight duration and image count
  - Mission metadata (date, operator, plot ID)
- Priority: High
- Dependencies: FR-PP-006, FR-PP-007

**FR-PP-010: Mission File Export**
- The system shall export mission files in JSON and/or XML formats
- The system shall save files with descriptive names including date and plot ID
- The system shall support export to user-specified directory
- Priority: High
- Dependencies: FR-PP-009

**FR-PP-011: Mission Preview and Validation**
- The system shall provide a summary report including:
  - Total waypoint count
  - Estimated flight time
  - Expected image count
  - Total flight distance
  - Battery requirements estimate
- The system shall display warnings if mission exceeds drone capabilities
- Priority: Medium
- Dependencies: FR-PP-009

### 4.2 Transfer Module

#### 4.2.1 Mission Folder Monitoring

**FR-TM-001: Real-Time Directory Monitoring**
- The system shall continuously monitor a user-configured mission folder for new image files
- The system shall detect new files within 10 seconds of creation
- The system shall support monitoring multiple subdirectories
- Priority: High
- Dependencies: None

**FR-TM-002: File Type Filtering**
- The system shall monitor only image file types: JPEG, PNG, TIFF, DNG
- The system shall ignore temporary, hidden, and system files
- Priority: High
- Dependencies: FR-TM-001

**FR-TM-003: Mission Session Detection**
- The system shall detect mission start based on:
  - First image timestamp
  - OpenPass mission log file presence (if available)
  - Manual trigger by operator
- The system shall group images by mission session
- Priority: Medium
- Dependencies: FR-TM-001

#### 4.2.2 Image Batching and Validation

**FR-TM-004: Intelligent Batching**
- The system shall batch images using configurable criteria:
  - Time window (e.g., every 5 minutes)
  - Image count threshold (e.g., every 50 images)
  - File size threshold (e.g., every 500MB)
  - End-of-mission trigger
- Priority: High
- Dependencies: FR-TM-001

**FR-TM-005: Image Quality Checks**
- The system shall validate each image for:
  - File integrity (readable, non-corrupted)
  - EXIF metadata presence (GPS coordinates, timestamp)
  - Minimum file size threshold
- The system shall quarantine invalid images and log errors
- Priority: High
- Dependencies: FR-TM-004

**FR-TM-006: Deduplication**
- The system shall compute hash (MD5 or SHA-256) for each image
- The system shall prevent upload of duplicate images based on hash comparison
- The system shall maintain a persistent deduplication database
- Priority: High
- Dependencies: FR-TM-005

**FR-TM-007: Batch Manifest Creation**
- The system shall create a manifest file for each batch including:
  - List of image filenames and paths
  - Hash values for each image
  - GPS coordinates from EXIF data
  - Timestamps and mission metadata
  - Batch ID and sequence number
- Priority: Medium
- Dependencies: FR-TM-006

#### 4.2.3 HPC Transfer and Job Triggering

**FR-TM-008: Tapis Authentication**
- The system shall authenticate with Tapis API using:
  - OAuth2 token-based authentication
  - Stored credentials with secure encryption
  - Automatic token refresh before expiration
- Priority: High
- Dependencies: None

**FR-TM-009: Asynchronous Upload**
- The system shall upload image batches to HPC storage asynchronously
- The system shall support concurrent upload of multiple batches (up to 3 simultaneous)
- The system shall not block new image monitoring during uploads
- Priority: High
- Dependencies: FR-TM-008

**FR-TM-010: Upload Resilience**
- The system shall implement retry logic with exponential backoff (max 5 retries)
- The system shall support partial upload recovery using chunked transfer
- The system shall resume interrupted uploads from last successful chunk
- The system shall queue batches locally if connectivity is unavailable
- Priority: Critical
- Dependencies: FR-TM-009

**FR-TM-011: Upload Verification**
- The system shall verify successful upload by:
  - Confirming file existence on HPC storage
  - Validating file sizes match local copies
  - Comparing hash values (if supported by Tapis)
- Priority: High
- Dependencies: FR-TM-009

**FR-TM-012: HPC Job Submission**
- The system shall submit HPC jobs via Tapis upon successful batch upload
- The system shall support configurable job types:
  - ML model training
  - Inference/prediction
  - Image preprocessing pipelines
  - Custom user-defined workflows
- Priority: High
- Dependencies: FR-TM-011

**FR-TM-013: Job Configuration**
- The system shall allow users to configure job parameters:
  - Job type and application ID
  - Compute resource allocation (nodes, cores, memory)
  - Maximum execution time
  - Input/output file paths
  - Environment variables and arguments
- Priority: Medium
- Dependencies: FR-TM-012

**FR-TM-014: Job Metadata Passing**
- The system shall pass batch manifest and mission metadata to HPC jobs
- The system shall include mission ID, plot ID, and image count in job context
- Priority: Medium
- Dependencies: FR-TM-012

#### 4.2.4 Monitoring and Status Reporting

**FR-TM-015: Transfer Status Tracking**
- The system shall track and display status for each batch:
  - Pending (queued locally)
  - Uploading (in progress)
  - Verifying (post-upload checks)
  - Completed (uploaded and verified)
  - Failed (error state with reason)
- Priority: High
- Dependencies: FR-TM-009

**FR-TM-016: Progress Reporting**
- The system shall display real-time upload progress:
  - Percentage complete for current batch
  - Transfer speed (MB/s)
  - Estimated time remaining
  - Queue depth for pending batches
- Priority: Medium
- Dependencies: FR-TM-015

**FR-TM-017: Logging and Notifications**
- The system shall log all significant events:
  - New image detection
  - Batch creation and upload initiation
  - Upload completion or failure
  - Job submission and status changes
  - Errors and warnings
- The system shall support configurable log levels (DEBUG, INFO, WARNING, ERROR)
- Priority: Medium
- Dependencies: None

**FR-TM-018: Error Alerting**
- The system shall provide alerts for critical failures:
  - Repeated upload failures (3+ consecutive)
  - Authentication errors
  - Storage capacity issues
  - Job submission failures
- The system shall support email or webhook notifications (optional)
- Priority: Medium
- Dependencies: FR-TM-017

### 4.3 System Integration and Configuration

**FR-SYS-001: Configuration Management**
- The system shall read configuration from YAML or JSON files
- The system shall support command-line argument overrides
- The system shall validate configuration on startup
- Priority: High
- Dependencies: None

**FR-SYS-002: User Configuration Parameters**

## 7. Appendix

### 7.1 Glossary

**Aerial Data Collection**: The process of capturing imagery or sensor data from unmanned aerial vehicles for analysis

**Asynchronous Upload**: Non-blocking file transfer that allows other system operations to proceed simultaneously

**Batch**: A grouped collection of images transferred as a single unit to optimize network efficiency

**Checkpoint**: A saved state marker enabling resumption of interrupted operations

**Chunked Transfer**: Breaking large files into smaller pieces for reliable transmission over unreliable networks

**Exponential Backoff**: Retry strategy where wait time increases exponentially between attempts

**Field Campaign**: A planned period of drone data collection operations at one or more sites

**GeoJSON**: JSON-based format for encoding geographic data structures

**Lawnmower Pattern**: Flight path consisting of parallel transects with 180° turns, resembling lawn mowing

**Nadir**: Directly downward viewing angle (0° from vertical)

**Quarantine**: Isolation of problematic files for manual review without affecting system operation

**Transect**: A straight-line flight segment along which data is collected

**Waypoint**: A specific geographic location (lat/lon/altitude) that defines a flight path vertex

## 2. Introduction

### 2.1 Purpose

The purpose of this Software Requirements Specification (SRS) is to document the complete requirements for the **Autonomous Aerial Data Collection System (AADC)**. This system aims to automate the process of aerial data acquisition, from generating efficient flight plans for Parrot Anafi drones based on user-defined plots to reliably transferring the collected imagery from edge devices to High-Performance Computing (HPC) systems for machine learning and other downstream processing. This document serves as a foundational agreement for development, testing, and acceptance.

### 2.2 Scope

The AADC system comprises two main components:

2.  **Transfer Module (TM):** Monitors a designated local mission folder for newly captured images, batches them, and manages robust, asynchronous uploads to remote HPC clusters (e.g., OSC) using Tapis or an equivalent reliable transfer mechanism. It also triggers subsequent HPC jobs (e.g., ML training/inference) upon successful data transfer.

The system will operate on a resource-constrained **Linux edge device** (e.g., a field laptop or NUC).
